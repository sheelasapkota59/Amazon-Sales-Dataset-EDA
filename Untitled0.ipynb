```python
import os
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Lambda
from tensorflow.keras.applications import ResNet50, EfficientNetB0
from tensorflow.keras.applications.resnet50 import preprocess_input as res_preprocess
from tensorflow.keras.applications.efficientnet import preprocess_input as eff_preprocess
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
from numpy.linalg import norm
import random
```

```python
IMG_SIZE = (224, 224)
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)
```

```python
train_folder = "/content/dataset/Animal Identification/train"
test_folder  = "/content/dataset/Animal Identification/test"

train_images = [os.path.join(train_folder, f) for f in os.listdir(train_folder)
                if f.lower().endswith(('.png','.jpg','.jpeg'))]
test_images  = [os.path.join(test_folder, f) for f in os.listdir(test_folder)
                if f.lower().endswith(('.png','.jpg','.jpeg'))]
```

```python
# Extract labels from filenames (assumes label is first word)
def get_label(path):
    return os.path.basename(path).split()[0]

train_labels = [get_label(p) for p in train_images]
test_labels  = [get_label(p) for p in test_images]

print(f"Train images: {len(train_images)}, Test images: {len(test_images)}")

# ---------------- LOAD PRETRAINED MODELS ----------------
res_base = ResNet50(weights='imagenet', include_top=False, pooling='avg', input_shape=(224,224,3))
res_model = Model(inputs=res_base.input, outputs=res_base.output)

eff_base = EfficientNetB0(weights='imagenet', include_top=False, pooling='avg', input_shape=(224,224,3))
eff_model = Model(inputs=eff_base.input, outputs=eff_base.output)

print("Pretrained models loaded!")
```

```python
def extract_features(img_path, model_type="resnet"):
    img = cv2.imread(img_path)
    img = cv2.resize(img, IMG_SIZE)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = img.astype(np.float32)
    if model_type=="resnet":
        img = res_preprocess(img)
        feat = res_model.predict(np.expand_dims(img, axis=0), verbose=0)
    else:
        img = eff_preprocess(img)
        feat = eff_model.predict(np.expand_dims(img, axis=0), verbose=0)
    return feat.flatten()

def cosine_similarity(vec1, vec2, eps=1e-8):
    return np.dot(vec1, vec2) / (max(norm(vec1)*norm(vec2), eps))
```

```python
def create_pairs(images, labels, other_images=None, num_pairs_per_class=50):
    """
    images: list of images for positive sampling
    labels: corresponding labels
    other_images: list of images from other set for negative sampling
    """
    pairs = []
    pair_labels = []

    unique_labels = list(set(labels))

    for label in unique_labels:
        # positive images
        pos_images = [img for img, lab in zip(images, labels) if lab == label]

        # negative images from other_images if provided
        neg_images = []
        if other_images is not None:
            neg_images = other_images

        # create positive pairs
        for _ in range(num_pairs_per_class):
            if len(pos_images) >= 2:
                i1, i2 = random.sample(pos_images, 2)
                pairs.append((i1, i2))
                pair_labels.append(1)

            # create negative pair only if neg_images is not empty
            if neg_images:
                i1 = random.choice(pos_images)
                i2 = random.choice(neg_images)
                pairs.append((i1, i2))
                pair_labels.append(0)

    return pairs, pair_labels

# TRAIN PAIRS: negative images are other train labels
train_pairs, y_train = create_pairs(train_images, train_labels, other_images=train_images, num_pairs_per_class=50)

# TEST PAIRS: negative images from train folder
test_pairs, y_test   = create_pairs(test_images, test_labels, other_images=train_images, num_pairs_per_class=50)
```

```python
# ---------------- EXTRACT FEATURES FOR PAIRS ----------------
def features_for_pairs(pairs, model_type="resnet"):
    x1, x2 = [], []
    for img1, img2 in pairs:
        x1.append(extract_features(img1, model_type))
        x2.append(extract_features(img2, model_type))
    return np.array(x1), np.array(x2)

x1_train_res, x2_train_res = features_for_pairs(train_pairs, "resnet")
x1_test_res, x2_test_res   = features_for_pairs(test_pairs, "resnet")

x1_train_eff, x2_train_eff = features_for_pairs(train_pairs, "efficient")
x1_test_eff, x2_test_eff   = features_for_pairs(test_pairs, "efficient")

y_train = np.array(y_train)
y_test  = np.array(y_test)
```

```python
# ---------------- BUILD SIAMESE MODEL ----------------
def build_siamese(input_shape):
    input_a = Input(shape=(input_shape,))
    input_b = Input(shape=(input_shape,))
    l1 = Lambda(lambda tensors: tf.abs(tensors[0] - tensors[1]))([input_a, input_b])
    output = Dense(1, activation='sigmoid')(l1)
    model = Model([input_a, input_b], output)
    model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
    return model

res_siamese = build_siamese(x1_train_res.shape[1])
eff_siamese = build_siamese(x1_train_eff.shape[1])

# ---------------- TRAIN ----------------
history_res = res_siamese.fit([x1_train_res, x2_train_res], y_train,
                              validation_data=([x1_test_res, x2_test_res], y_test),
                              epochs=10, batch_size=16, verbose=1)

history_eff = eff_siamese.fit([x1_train_eff, x2_train_eff], y_train,
                              validation_data=([x1_test_eff, x2_test_eff], y_test),
                              epochs=10, batch_size=16, verbose=1)
```

```python
# ---------------- EVALUATE ----------------
def evaluate(model, x1, x2, y, title="Model"):
    y_pred = model.predict([x1, x2])
    y_pred_labels = (y_pred > 0.5).astype(int)
    acc = accuracy_score(y, y_pred_labels)
    prec = precision_score(y, y_pred_labels)
    rec = recall_score(y, y_pred_labels)
    f1 = f1_score(y, y_pred_labels)
    print(f"\n{title} Evaluation: Accuracy={acc:.2f}, Precision={prec:.2f}, Recall={rec:.2f}, F1={f1:.2f}")
    cm = confusion_matrix(y, y_pred_labels)
    disp = ConfusionMatrixDisplay(cm, display_labels=["Not Similar","Similar"])
    disp.plot(cmap='Blues')
    plt.title(f"{title} Confusion Matrix")
    plt.show()
    return y_pred_labels

y_pred_res = evaluate(res_siamese, x1_test_res, x2_test_res, y_test, "ResNet50 Siamese")
y_pred_eff = evaluate(eff_siamese, x1_test_eff, x2_test_eff, y_test, "EfficientNetB0 Siamese")
```

```python
def plot_training_history(history, title="Training vs Validation"):
    plt.figure(figsize=(8,6))
    plt.plot(history.history['accuracy'], label='Train Accuracy', marker='o')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy', marker='o')
    plt.plot(history.history['loss'], label='Train Loss', marker='x')
    plt.plot(history.history['val_loss'], label='Validation Loss', marker='x')
    plt.title(title)
    plt.xlabel('Epochs')
    plt.ylabel('Value')
    plt.legend()
    plt.grid(True)
    plt.show()

# Figure 9: ResNet50
plot_training_history(history_res, "Figure 9: Training vs Validation Accuracy and Loss (ResNet50)")

# Figure 10: EfficientNetB0
plot_training_history(history_eff, "Figure 10: Training vs Validation Accuracy and Loss (EfficientNetB0)")
```
